## Decision Structures and Boolean Logic

Two basic types of control structures:

1. Sequence structure
      * Set of statements that execute in the order they appear.
2. Decision structure (selection structure)
      * Specific action(s) performed only if a condition exists.
      * Flowchart is a useful tool to visualize the decision structure.

## 1. The <i>if</i> Statement

**Note:** The `if` statement is used to create a decision structure, which allows a program to have **more than one** path of execution. The general format of the if statement:

`if` condition:
> statement
>
> statement
>
> etc.

````python
if [condition]:
    statement1
    statement2
    statement3

statement4
````

**Note:** Pay attention to following points:

* The word `if` is followed by a `condition`, which is an expression that will be evaluated as either `True` or `False`.
* A **colon** appears after the `condition`.
* [Important] The next line begins a block of statements.
  * A block is a set of statements that belong together as a group.
  * All statements in the block are **indented** to indicate where the block begins and ends.

**Note:** When the `if` statement executes, the condition is tested.

* If the condition is `True`, the statements in the block following the if clause are executed.
* If the condition is `False`, the statements in the block are skipped.

**Example:** Chris owns an auto repair business and has several employees. If any employee works over 40 hours in a week, he pays them 1.5 times their regular hourly pay rate for all houurs over 40. He has asked you to design a simple payroll program that calculates an employee's gross pay, including any overtime wages.

````python
# Define constant variables
G_WORK_HOUR = 40
OT_RATE = 1.5

# Get the work hour of the employee
work_hour = int(input('How many hours did you work?'))

#Get the pay rate
pay_rate = float(input('What is the pay rate per hour?'))

if work_hour>G_WORK_HOUR:
    ot_work_hour =  work_hour - G_WORK_HOUR
    ot_payment = ot_work_hour * pay_rate * OT_RATE
    regular_payment = G_WORK_HOUR * pay_rate
    gross_payment = regular_payment + ot_payment
else:
    regular_payment = work_hour * pay_rate
    gross_payment = regular_payment

print(f'Here is the gross payment amount :${gross_payment}')
````

## 4. Nested Decision Structures and the <i>if-elif-else</i> Statement

**Note:** To test more than one condition, a decision structure can be nested inside another decision structure.

**Example:**

* Determine if someone qualifies for a loan, they must meet two conditions:
  * Must earn at least $30,000/year
  * Must have been employed for at least two years
* Check first condition, and if it is `True`, check second condition

<img src='https://kambizsaffari.com/SC/3_12.png' width='700px' />

*Source: Starting Out with Python (5ed)- by Tony Gaddis Pearson*

**Example:**

````python
# Assume the minimum annual salary and the minimum years on the job are as fallows:
MIN_SALARY = 30000
MIN_YEARS = 2

# Get the customer's annual salary.
salary = float(input('Enter your annual salary: '))

# Get the number of years on the current job.
years_on_job = int(input('Enter the number of years employed: '))

# Determine whether the customer qualifies. if salary >= MIN_SALARY:
if salary >= MIN_SALARY:
    if years_on_job >= MIN_YEARS:
        print('You qualify for the loan.')
    else:
        print('You must have been employed for at least', MIN_YEARS, 'years to qualify.')
else:
    print('You must earn at least $',format(MIN_SALARY, ',.2f'),' per year to qualify.', sep='')
````

**Note:** Important to use proper indentation in a nested decision structure.

* Important for Python interpreter
* Makes code more readable for programmer
* Follow these rules when writing nested `if` statements:
  * Each else clause is aligned with its matching `if` clause.
  * The statements in each block are consistently indented.
  * Notice each statement in each block is indented the same amount.

### 4.1. the <i>if-elif-else</i> Statement

`if` condition_1:
> statement
>
> statement
>
> etc.

> statement
>
> statement
>
> etc.

*Insert as many elif clauses as necessary . . .*

`else`:
> statement
>
> statement
>
> etc.

**Note:** Alignment used with `if-elif-else` statements:

* The `if`, `elif`, and `else` clauses are all aligned.
* Conditionally executed blocks are consistently indented.

The `if-elif-else` statement is not strictly required because its logic can be implemented using nested `if-else` statements.

**However,** the advantages of using the `if-elif-else` statement instead are:

* It is easier to follow than a long series of nested `if-else` statements.
* The lines of code tend to be shorter.

**Note:** The `not` operator takes a Boolean expression as its operand and reverses its logical value.

* If the expression is `True`, the not operator returns `False`.

```

    print('This is below the maximum temperature.')
```
